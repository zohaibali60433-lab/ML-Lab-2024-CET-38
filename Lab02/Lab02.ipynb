**Task 6:**

```python
# Task no. 6
areas = np.array(["Punjab", "Sindh", "KP", "Balochistan", "Islamabad", "AJK"])
pop17 = np.array([110.01, 47.85, 35.53, 12.34, 2.01, 4.05])
pop2023 = np.array([127.69, 55.70, 40.85, 14.89, 2.36, 4.05])

def asDict(pairs):
    out = {}
    for kIdx, elem in pairs:
        out[str(kIdx)] = elem.item() if hasattr(elem, "item") else elem
    return out

grandTotal = np.sum(pop2023)
print(f"#6a: total population 2023 = {grandTotal:.2f} million")

shareOfTotal = (pop2023 / grandTotal) * 100
print(f"#6b: share of total (%): {asDict(zip(areas, np.round(shareOfTotal, 2)))}")

print(f"#6c: most populous unit: {areas[pop2023.argmax()]}")
print(f"#6d: units above 20 million: {areas[pop2023 > 20]}")

growth = 100 * ((pop2023 / pop17) ** (1 / 6) - 1)
print(f"#6e: annual growth 2017-2023 (% per year): {asDict(zip(areas, np.round(growth, 2)))}")

print(f"#6f: fastest-growing unit: {areas[np.argmax(growth)]} ({growth.max():.2f}% per year)")

natGrowth = ((pop2023.sum() / pop17.sum()) ** (1 / 6) - 1) * 100
doubleYrs = 70 / growth.max()
print(f"#6 interpret: national growth = {natGrowth:.2f}% per year.")
print(f"#6 interpret: {areas[np.argmax(growth)]} would double in about {doubleYrs:.0f} years at its current rate.")
print(f"#6 interpret: AJK grew {growth[-1]:.2f}% per year vs national {natGrowth:.2f}% -> "
      f"AJK is {natGrowth - growth[-1]:.2f} points below the national rate (in this sample AJK is essentially flat).")
```

**Task 7:**

```python
# Task no. 7
cityList = np.array(["Lahore", "Karachi", "Islamabad", "Quetta", "Peshawar"])
monNames = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                   "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])

temp = np.array([
    [19, 22, 28, 35, 40, 40, 36, 34, 34, 32, 26, 21],
    [26, 28, 32, 34, 35, 34, 32, 31, 32, 34, 31, 28],
    [17, 19, 24, 30, 36, 38, 35, 33, 33, 30, 24, 19],
    [11, 13, 18, 24, 29, 34, 35, 34, 31, 25, 19, 13],
    [18, 20, 25, 31, 37, 40, 38, 36, 35, 31, 25, 19],
])
rainfall = np.array([
    [12, 22, 34, 22, 21, 54, 214, 232, 82, 8, 3, 8],
    [8, 10, 9, 4, 1, 6, 40, 25, 13, 1, 2, 5],
    [56, 65, 76, 52, 32, 66, 260, 310, 120, 28, 12, 28],
    [38, 40, 33, 20, 8, 5, 18, 15, 5, 3, 7, 18],
    [33, 48, 63, 48, 23, 21, 86, 95, 43, 23, 20, 29],
])

def asDict(pairs):
    out = {}
    for kIdx, elem in pairs:
        out[str(kIdx)] = elem.item() if hasattr(elem, "item") else elem
    return out

print(f"#7a: shape of temp (cities, months): {temp.shape}")

meanYearly = np.mean(temp, axis=1)
print(f"#7b: yearly mean temperature (deg C): {asDict(zip(cityList, np.round(meanYearly, 1)))}")

hottestMon = monNames[np.argmax(temp, axis=1)]
print(f"#7c: hottest month per city: {asDict(zip(cityList, hottestMon))}")

print(f"#7d: 5-city monthly average (deg C): {asDict(zip(monNames, np.round(temp.mean(axis=0), 1)))}")

count35 = (temp > 35).sum(axis=1)
print(f"#7e: months above 35 deg C: {asDict(zip(cityList, count35))}")

lahoreF = temp[0] * 9 / 5 + 32
print(f"#7f: Lahore monthly temperature (deg F): {np.round(lahoreF, 1)}")

monsoonRain = np.sum(rainfall[:, 6:9], axis=1)
print(f"#7g: monsoon rain Jul-Sep (mm): {asDict(zip(cityList, monsoonRain))}")
print(f"#7g: wettest monsoon city: {cityList[np.argmax(monsoonRain)]} ({monsoonRain.max()} mm)")

shareOfTotal = monsoonRain / rainfall.sum(axis=1) * 100
print(f"#7h: monsoon share of annual rain (%): {asDict(zip(cityList, np.round(shareOfTotal, 1)))}")

kIdx = np.flatnonzero(cityList == "Karachi")[0]
print(f"#7 interpret: Karachi has the highest yearly mean ({meanYearly.max():.1f} deg C) "
      f"but {count35[kIdx]} months above 35 deg C, while cooler-on-average Peshawar has {count35[4]} "
      f"and Lahore {count35[0]}. A high yearly mean does not mean the most extreme heat: "
      f"Karachi is hot all year, the inland cities spike in summer.")
```

**Task 8:**

```python
# Task no. 8
fyLabels = np.array(["FY2023", "FY2024", "FY2025"])
fmonths = np.array(["Jul", "Aug", "Sep", "Oct", "Nov", "Dec",
                    "Jan", "Feb", "Mar", "Apr", "May", "Jun"])

inflation = np.array([
    [24.9, 27.3, 23.2, 26.6, 23.8, 24.5, 27.6, 31.5, 35.4, 36.4, 38.0, 29.4],
    [28.3, 27.4, 31.4, 26.9, 29.2, 29.7, 28.3, 23.1, 20.7, 17.3, 11.8, 12.6],
    [4.1, 9.6, 6.9, 7.2, 4.9, 4.1, 2.4, 1.5, 0.7, 0.3, 3.5, 3.2],
])

def asDict(pairs):
    out = {}
    for kIdx, elem in pairs:
        out[str(kIdx)] = elem.item() if hasattr(elem, "item") else elem
    return out

meanInfl = np.mean(inflation, axis=1)
print(f"#8a: average inflation per year (%): {asDict(zip(fyLabels, np.round(meanInfl, 2)))}")

inflFy24 = inflation[1]
print(f"#8b: FY2024 months above 25 %: {fmonths[inflFy24 > 25]} -> {(inflFy24 > 25).sum()} months")

monthlyChange = np.diff(inflation, axis=1)
print(f"#8c: month-on-month change (percentage points; negative = drop):\n{np.round(monthlyChange, 1)}")

biggest = monthlyChange.argmin(axis=1)
print(f"#8d: month of biggest drop: {asDict(zip(fyLabels, fmonths[biggest + 1]))} | size (pp): {np.round(monthlyChange.min(axis=1), 1)}")

infl25 = inflation[2]
level = np.where(infl25 < 5, "Low", np.where(infl25 < 10, "Moderate", "High"))
print(f"#8e: FY2025 labels (Low < 5 %, Moderate 5-10 %, High >= 10 %):\n{np.column_stack((fmonths, infl25, level))}")

print(f"#8 interpret: average inflation went {meanInfl[0]:.1f}% -> {meanInfl[1]:.1f}% -> {meanInfl[2]:.1f}%. "
      f"Prices did NOT fall; they kept rising, but more and more slowly (disinflation), "
      f"so the price LEVEL is still much higher than at the start of FY2023.")
```

**Task 9:**

```python
# Task no. 9
marksArr = np.array([34, 38, 35, 52, 67, 73, 88, 95, 98, 39,
                  41, 45, 36, 58, 62, 71, 80, 33, 99, 44])
rollTags = np.arange(1, 21)
PASS = 40

marksAfter = np.clip(5 + marksArr, 0, 100)
print(f"#9a: original marks: {marksArr}")
print(f"#9a: marks after +5 grace (cap 100): {marksAfter}")

pctBefore = (marksArr >= PASS).mean() * 100
pAfter = (marksAfter >= PASS).mean() * 100
print(f"#9b: pass % before grace = {pctBefore:.1f} %, after grace = {pAfter:.1f} %")

justGrace = (marksArr < PASS) & (marksAfter >= PASS)
print(f"#9c: passed only because of grace (roll numbers): {rollTags[justGrace]}")
print(f"#9c: their original marks: {marksArr[justGrace]}")

z = (marksAfter - np.mean(marksAfter)) / np.std(marksAfter)
print(f"#9d: mean = {marksAfter.mean():.2f}, std = {marksAfter.std():.2f}")
print(f"#9d: z-scores (std devs from mean): {np.round(z, 2)}")
```

**Task 10:**

```python
# Task no. 10
img = np.zeros((8, 8), dtype=np.uint8)
img[1, 1:-1] = 255
img[[2, 3, 4, 5, 6], [6, 5, 5, 4, 4]] = 255

def asBits(a):
    return (a > 0).astype(int)

print(f"#10a: the '7' (1 = white):\n{asBits(img)}")

flipLr = np.fliplr(img)
print(f"#10b: flipped left-right:\n{asBits(flipLr)}")

invertImg = ~img
print(f"#10c: inverted (white <-> black):\n{asBits(invertImg)}")

cropped = img[0:7, 1:8]
print(f"#10d: cropped 7x7 region, shape: {cropped.shape}\n{asBits(cropped)}")

brightened = np.clip(img.astype(int) + 60, 0, 255).astype(np.uint8)
print(f"#10e: brightened (+60, capped at 255); unique values: {np.unique(brightened)}")

print(f"#10f: shape = {img.shape} | dtype = {img.dtype} | white pixels = {np.count_nonzero(img == 255)}")
```
