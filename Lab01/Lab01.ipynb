## Task 07

```python
def get_domain(email):
    return email.split("@")[-1]
```

```python
print("Q7:", get_domain("user@domain.com"))
```

## Task 08
Function: return True if the word "dog" is in a string (case-insensitive).

```python
def find_dog(text):
    return "dog" in text.lower()
```

```python
print("Q8:", find_dog("Is there a dog here?"))
print("Q8:", find_dog("Is there a DOG here?"))
print("Q8:", find_dog("No pets here"))
```

## Task 09
Function: count how many times "dog" occurs in a string (case-insensitive).

```python
def count_dog(text):
    return text.lower().count("dog")
```

```python
print("Q9:", count_dog("This dog runs faster than the other dog, DOG!"))
```

## Task 10
`caught_speeding(speed, is_birthday)`: ≤ 60 "No Ticket", 61–80 "Small Ticket", ≥ 81 "Big Ticket". On a birthday, allow 5 mph more.

```python
def caught_speeding(speed, is_birthday):
    allowance = 5 if is_birthday else 0
    if speed <= 60 + allowance:
        return "No Ticket"
    elif speed <= 80 + allowance:
        return "Small Ticket"
    else:
        return "Big Ticket"
```

```python
print("Q10:", caught_speeding(60, False))  
print("Q10:", caught_speeding(65, False))  
print("Q10:", caught_speeding(65, True))   
print("Q10:", caught_speeding(85, True))   
print("Q10:", caught_speeding(86, True))    
```
